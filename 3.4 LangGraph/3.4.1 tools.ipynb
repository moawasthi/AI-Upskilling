```python
from langchain_core.tools import tool
```

```python
def multiply(a:int, b: int) ->int:
    return a*b
```

```python
multiply(a=10, b = 20) # not a langchain tool yet
```

```python
@tool # decorator for langchain
def multiply(a:int, b: int) ->int:
    '''This function multiplies provided values together'''
    return a*b
```

```python
multiply.invoke({"a": 10, "b" : 20}) # now a langchain tool
```

```python
multiply.description
```

```python
multiply.args
```

```python
@tool
def getwordlength( word : str) -> int:
    '''This function returns the length of the word given'''
    return len(word)
```

```python
getwordlength.invoke({"word" : "Learning Langgraph custom tools"})
```
